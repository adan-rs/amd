## El curso como un proyecto de analítica

Un proyecto de analítica no empieza en el modelo: empieza en el negocio. Este curso replica deliberadamente las fases típicas de un proyecto real (basadas en la metodología **CRISP-DM**, el estándar más utilizado en la industria):

1. **Comprensión del negocio**: ¿qué pregunta queremos responder y para qué decisión?
2. **Comprensión de los datos**: ¿qué datos tenemos y qué representa cada variable?
3. **Preparación de los datos**: limpieza, datos faltantes, atípicos, estructura.
4. **Análisis y modelado**: desde la exploración descriptiva hasta las pruebas de hipótesis y los modelos estadísticos.
5. **Evaluación e interpretación**: ¿los resultados son sólidos? ¿son relevantes para el negocio?
6. **Comunicación**: traducir los hallazgos a un reporte que alguien pueda usar para decidir.

En la práctica profesional estas fases no son lineales (se regresa a preparar datos, se replantea la pregunta), pero el orden del curso las recorre una a una. La fase 6 está presente en *todos* los notebooks a través de los **ejemplos de reporte de resultados**, que muestran cómo redactar cada hallazgo.

**Lo que este curso no cubre (a propósito)**: los modelos de pronóstico (series de tiempo) y de clasificación (regresión logística, árboles, etc.) se abordan en materias posteriores del programa. Aquí construyes la base estadística y de criterio sobre la que esas materias se apoyan.

Los notebooks siguen dos formatos distintos, y la diferencia es intencional:

## Hoja de decisión: ¿qué técnica necesito?

### Paso 1. Identifica tu objetivo

| Si tu objetivo es... | La familia de técnicas es... | Notebook |
|----------------------|------------------------------|----------|
| Describir y explorar los datos | Medidas numéricas, visualización, tablas cruzadas, matriz de correlaciones | 05, 06, 09 |
| Representar visualmente la relación entre categorías (marcas, atributos) | Mapas perceptuales (análisis de correspondencias) | 12 |
| Probar si existe una diferencia o relación (con evidencia estadística) | Pruebas de hipótesis | 08, 10, 11, 14, 15 |
| Explicar o estimar una variable a partir de otras | Regresión lineal múltiple | 16, 17, 18 |
| Agrupar observaciones similares (clientes, productos) | Análisis de conglomerados | 19 |
| Resumir muchas variables correlacionadas en pocas dimensiones | PCA / análisis factorial | 20, 21 |

### Paso 2. Si vas a probar una hipótesis: ¿qué variables tienes?

- **¿Relación entre dos variables?**
    - Ambas *cualitativas* → **Chi cuadrada** (o exacta de Fisher si hay frecuencias esperadas menores a 5; nb. 10)
    - Ambas *cuantitativas* → **Correlación de Pearson** (o **Spearman** si hay no normalidad, datos ordinales o atípicos; nb. 11)
- **¿Comparación entre grupos?**
    - 2 grupos, comparando una *proporción* → **Prueba z de dos proporciones** (nb. 14)
    - 2 grupos, comparando una *media* → **Prueba t** (verifica varianzas con Levene; si difieren, **t de Welch**; si no hay normalidad y la muestra es pequeña, **Mann-Whitney**; nb. 14)
    - 3 o más grupos, comparando *medias* → **ANOVA** (si es significativo, post-hoc de Tukey/Bonferroni; si fallan los supuestos, **Welch ANOVA** o **Kruskal-Wallis**; nb. 15)
- **¿Necesitas verificar normalidad antes?** → Gráfico Q-Q + **Shapiro-Wilk** o **Kolmogorov-Smirnov-Lilliefors** (nb. 08)

### Paso 3. Si vas a reducir dimensiones o agrupar

- Agrupar *observaciones*: pocas observaciones y quieres visualizar la estructura → **método jerárquico** (dendrograma); muchas observaciones → **k-medias** (explora k con silueta o codo; nb. 19)
- Variables numéricas correlacionadas y quieres *índices no correlacionados* → **PCA** (nb. 20)
- Ítems de encuesta que miden *constructos latentes* (actitudes, percepciones) → **Análisis factorial** + alfa de Cronbach (nb. 21)

### Recordatorios transversales

- **Estandariza las variables** antes de aplicar PCA, análisis factorial o clusterización.
- En toda prueba de hipótesis: p-valor < α (usualmente 0.05) → se rechaza $H_0$; pero *no rechazar* no demuestra que $H_0$ sea cierta (nb. 07).
- **Significativo no es lo mismo que importante**: reporta siempre la magnitud del efecto en unidades del negocio, no solo el p-valor (nb. 07).
- Antes de cualquier técnica: comprende y prepara los datos (nb. 03 y 04). Ningún modelo compensa datos mal preparados.

# Proyecto final

El proyecto final es la pieza que integra todo el curso: en equipo, tomarán un conjunto de datos real y recorrerán **todas las fases de un proyecto de analítica** (del planteamiento de la pregunta al reporte final) aplicando las técnicas vistas en las sesiones.

## Modalidad y rutas de trabajo

- **Equipos de hasta 3 integrantes.** Se recomienda combinar perfiles: alguien con experiencia en negocios y alguien con más soltura en programación aprenderán más juntos que separados.
- Cada equipo elige **una de dos rutas**, y esa elección define el enfoque del proyecto y el formato del reporte final:

**Ruta 1. Proyecto aplicado en su empresa o entorno profesional.** El objetivo es resolver un problema real con los datos disponibles, generando valor práctico para la organización: mejorar procesos mediante análisis de causas, entender comportamientos de clientes mediante segmentación o mapas perceptuales, optimizar ventas, operaciones o recursos con modelos de regresión o series de tiempo. El proyecto se articula alrededor de una **pregunta de negocio** y cierra con un reporte ejecutivo.

**Ruta 2. Análisis académico con enfoque riguroso.** El objetivo es realizar un análisis que cumpla con estándares académicos, como si fuera a presentarse en un congreso de investigación. Puede basarse en bases de datos públicas, encuestas u otros instrumentos válidos. Se espera una **pregunta de investigación** clara, justificación teórica y aplicación metodológica sólida, y cierra con un reporte académico breve.

Ambas rutas comparten el mismo alcance analítico y la misma rúbrica; lo que cambia es el marco de la pregunta y el formato del entregable final. En adelante, donde se lea *pregunta de negocio o de investigación*, cada equipo debe entender la que corresponda a su ruta.

## Dataset

**Dataset libre, con aprobación del profesor** (Hito 1). Debe provenir de una fuente oficial o institucional: los datos de su propia organización (anonimizados), portales de datos abiertos de gobierno u organismos públicos (INEGI, data.gob.mx, Banco Mundial, CONEVAL), bases de organismos internacionales o de instituciones académicas, o encuestas e instrumentos aplicados por ustedes mismos. **No se aceptan datasets descargados de repositorios de práctica como Kaggle, UCI o similares**, ni datos sin una fuente identificable y verificable. En el Hito 1 el profesor aprueba tanto el dataset como el planteamiento del proyecto, o sugiere ajustes; no avancen al Hito 2 sin esa aprobación.

**Requisitos mínimos del dataset**:
- **Datos reales y de fuente identificable**, no sintéticos ni simulados. Pueden estar razonablemente limpios; lo que no se acepta es un dataset generado artificialmente para el ejercicio, ni uno cuya procedencia no pueda citarse.
- Suficientes observaciones y variables para sostener el alcance mínimo descrito abajo, con **mezcla de variables cuantitativas y cualitativas**.
- Debe permitir plantear una **pregunta de negocio o de investigación** cuya respuesta apoye una decisión o aporte conocimiento (¿qué segmentos existen?, ¿difiere X entre grupos?, ¿qué factores explican Y?).

Si el dataset trae valores faltantes, atípicos o inconsistencias, mejor: la preparación de datos es parte del proyecto y se califica.

## Alcance mínimo

El proyecto debe incluir, como mínimo:

1. **Planteamiento**: pregunta de negocio o de investigación, contexto, justificación del valor del análisis y la decisión o el conocimiento que se apoyaría con él.
2. **Preparación de datos**: revisión de calidad y tratamiento justificado de faltantes, duplicados y atípicos (notebooks 03-04).
3. **Análisis exploratorio**: medidas descriptivas y al menos tres visualizaciones con interpretación (notebooks 05-06).
4. **Análisis principal**: al menos una **técnica de complejidad media** aplicada correctamente, con supuestos verificados e interpretación sustantiva. Se consideran adecuadas, entre otras: regresión múltiple o logística, ANOVA con más de un factor o con contrastes post hoc, análisis de clúster, análisis factorial o de componentes principales, mapas perceptuales, series de tiempo, árboles de decisión y otros modelos de clasificación (notebooks 07-21).

   No se considera suficiente un proyecto que se agote en estadística descriptiva o en una prueba bivariada simple (un coeficiente de correlación, una comparación de dos medias). Estas pruebas pueden y suelen formar parte del recorrido, pero el análisis debe llegar a una técnica que ponga varias variables en juego y permita responder la pregunta con profundidad. Aplicar más de una técnica es válido y suma en el criterio de criterio analítico.

5. **Comunicación**: reporte final, en el formato que corresponda a su ruta (ver abajo).

## Hitos y calendario

Los hitos son entregas cortas cuyo propósito es avanzar a tiempo y recibir retroalimentación; cada uno vale el 10% de la calificación final del curso.

| Hito | Sesión | Entregable | Peso |
|------|--------|-----------|------|
| **1. Definición del proyecto** | Sesión 2 | Documento con integrantes y ruta elegida, título tentativo, descripción del problema o pregunta, justificación del valor del análisis, objetivo general y específicos, tipo de datos y cómo se obtendrán, y técnicas de análisis tentativas. El profesor aprueba el proyecto y el dataset, o sugiere ajustes. | 10% |
| **2. Preparación de datos y estadística descriptiva** | Sesión 4 | Documento con la evidencia de los datos, un primer análisis descriptivo y los comentarios sobre lo observado. | 10% |

Los hitos se entregan en la plataforma en la fecha indicada. Un hito entregado a tiempo pero con áreas de mejora recibe retroalimentación sin penalización mayor; un hito no entregado sí pierde su peso.

## Hito 1. Definición del proyecto (Sesión 2)

Un documento por equipo que incluya:

1. Integrantes y **ruta elegida** (aplicada o académica).
2. Título tentativo del proyecto.
3. Descripción del problema o pregunta de investigación.
4. Justificación del valor del análisis (para la empresa o para el conocimiento académico).
5. Objetivo general y objetivos específicos.
6. Tipo de datos que se utilizarán, su fuente y cómo se obtendrán, con una nota sobre la **factibilidad** en tiempo, recursos y acceso a los datos.
7. Técnicas de análisis tentativas.

El profesor responderá aprobando el proyecto o pidiendo ajustes. La aprobación del dataset es requisito para continuar.

## Hito 2. Preparación de datos y estadística descriptiva (Sesión 4)

En este segundo avance, cada equipo deberá mostrar evidencia de que ya cuenta con los datos de su proyecto y de que ha comenzado a trabajar con ellos.

El documento debe indicar de dónde provienen los datos y qué abarcan, describir brevemente la base con la que se está trabajando, y presentar un primer análisis descriptivo con las medidas y gráficas que el equipo considere más pertinentes para su caso (no se exige un conjunto específico).

Se debe cerrar con comentarios sobre lo que se observa en los datos hasta el momento y cómo se relaciona con la pregunta y los objetivos del primer avance, señalando cualquier dificultad o ajuste que haya surgido. La entrega es en archivo PDF o de Word, en la fecha indicada en la plataforma.

**Rúbrica general** (los porcentajes se reparten dentro del 10% de este avance):

- **Evidencia de los datos (30%)**: se aporta la descripción de la base de datos y se explica claramente su origen, cobertura y contenido.
- **Análisis descriptivo (30%)**: se presentan medidas de resumen o gráficas pertinentes al tipo de variables y al objetivo del proyecto.
- **Interpretación (30%)**: se comentan los hallazgos preliminares y se vinculan con la pregunta y los objetivos del proyecto.
- **Presentación y entrega (10%)**: documento ordenado, con redacción clara, y entregado completo y en tiempo.

## Entrega final (sesión 8)

Dos componentes:

1. **Notebook completo** (ipynb exportado a html o pdf): todo el flujo, con código documentado y decisiones justificadas. Debe poder leerse de principio a fin como la historia del proyecto. Es el mismo requisito para ambas rutas.

2. **Reporte final** (pdf, **sin código**), en el formato de su ruta:

   - **Ruta 1. Reporte ejecutivo, máximo 2 cuartillas** sin contar portada, referencias ni anexos. Dirigido a una audiencia directiva: pregunta, hallazgos principales en lenguaje de negocio, recomendación y limitaciones. Los *ejemplos de reporte de resultados* de los notebooks del curso son la referencia de estilo.
   - **Ruta 2. Reporte académico, máximo 4 cuartillas** sin contar portada, referencias ni anexos. Estructura de artículo breve: introducción y justificación, breve marco de referencia, metodología (datos, variables y técnicas), resultados, discusión y limitaciones, y referencias en un formato de citación consistente (APA u otro estándar).

Los límites son máximos, no metas: un reporte más corto que cumpla su función está bien. Las tablas y gráficas que van dentro del cuerpo cuentan para el límite; las que se mandan al anexo, no. Los anexos son material de respaldo y no sustituyen al reporte: lo que se califica es el cuerpo, que debe entenderse sin abrirlos. Si por la naturaleza del proyecto necesitan más extensión, pueden plantearlo al profesor antes de la entrega: los casos excepcionales se aprueban de manera individual.

No hay presentación: el peso de la comunicación recae en el reporte, que debe sostenerse por sí solo sin que nadie lo explique. Escríbanlo pensando en un lector que solo tiene esas cuartillas.

**Peso del proyecto en la calificación final del curso**

| Hito 2. Preparación de datos y estadística descriptiva | 10% |

| **Total del proyecto** | **40%** |

**Rúbrica de la entrega final (20% de la nota final)**

Los porcentajes de los criterios siguientes se reparten dentro de ese 20%.

1. **Comprensión del problema y planteamiento (10%)**: pregunta clara, relevante y conectada con una decisión o con una contribución al conocimiento.
2. **Preparación y calidad de datos (15%)**: tratamiento técnicamente correcto y justificado; decisiones documentadas.
3. **Análisis exploratorio e inferencial (15%)**: pruebas apropiadas para el tipo de variables, supuestos verificados, interpretación correcta de p-valores y magnitudes.
4. **Análisis principal: modelado o técnica multivariada (20%)**: técnica bien seleccionada y aplicada, con evaluación de supuestos o calidad de la solución, e interpretación sólida.
5. **Comunicación: reporte final (20%)**: narrativa clara y hallazgos bien traducidos dentro del límite de extensión (a lenguaje de negocio con una recomendación accionable en la Ruta 1; a un argumento metodológicamente sustentado y bien referenciado en la Ruta 2).
6. **Criterio analítico y limitaciones (10%)**: decisiones razonadas a lo largo del flujo y reconocimiento honesto de las limitaciones del análisis.
7. **Extensión del análisis con IA Generativa (10%)**: uso de la IA para ir más allá de lo solicitado, con un anexo que incluya los *prompts* utilizados y una reflexión crítica sobre las respuestas obtenidas.

*Igual que en las prácticas: un proyecto que "corre" pero no interpreta, o que interpreta sin sustento en los datos, no alcanzará la calificación máxima. Significativo no es lo mismo que importante: la conclusión final debe considerar la relevancia práctica, no solo la estadística.*

## Errores comunes (evítenlos)

- **Empezar el proyecto en la última semana.** Los hitos existen precisamente para evitarlo.
- **Avanzar sin la aprobación del Hito 1.** Un dataset o un planteamiento no aprobados suelen obligar a rehacer trabajo.
- **Quedarse en lo simple.** Un proyecto que termina en descriptivos y una correlación no cumple el alcance mínimo.
- **Aplicar técnicas sin pregunta.** Primero la pregunta, luego la técnica.
- **Reportar salidas de Python sin interpretación**, o llenar el reporte final de tablas y código.
- **Ignorar los supuestos** de las pruebas y modelos: verificar supuestos es parte de la calificación.
- **Delegar el análisis a la IA.** Úsenla para profundizar y contrastar; las conclusiones deben reflejar comprensión propia y coherencia con sus propios resultados.

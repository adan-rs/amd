# Proyecto final: un proyecto de analítica de principio a fin

El proyecto final es la pieza que integra todo el curso: en equipo, tomarán un conjunto de datos real y recorrerán **todas las fases de un proyecto de analítica** —de la pregunta de negocio al reporte ejecutivo— aplicando las técnicas vistas en las sesiones. No es un trabajo adicional al curso: es el curso, aplicado a un problema que ustedes elijan.

## Modalidad y dataset

- **Equipos de 2 a 3 personas**. Se recomienda combinar perfiles: alguien con experiencia en negocios y alguien con más soltura en programación aprenderán más juntos que separados.
- **Dataset libre con aprobación del profesor** (Hito 1). Puede provenir de su trabajo (anonimizado), de portales de datos abiertos (INEGI, data.gob.mx, Inside Airbnb), de Kaggle o de otra fuente pública.

**Requisitos mínimos del dataset**:
- Al menos **200 observaciones** y **8 variables**, con mezcla de variables cuantitativas y cualitativas.
- Datos **reales e "imperfectos"**: un dataset con valores faltantes, atípicos o inconsistencias es deseable, porque la preparación de datos es parte del proyecto. Eviten datasets sintéticos o ya perfectamente limpios.
- Debe permitir plantear una **pregunta de negocio** cuya respuesta apoye una decisión (¿qué segmentos existen?, ¿difiere X entre grupos?, ¿qué factores explican Y?).

## Alcance mínimo

El proyecto debe incluir, como mínimo:

1. **Planteamiento**: pregunta de negocio, contexto y decisión que se apoyaría con el análisis.
2. **Preparación de datos**: revisión de calidad, tratamiento justificado de faltantes, duplicados y atípicos (notebooks 03-04).
3. **Análisis exploratorio**: medidas descriptivas y al menos tres visualizaciones con interpretación (notebooks 05-06).
4. **Análisis inferencial**: al menos **dos pruebas de hipótesis de familias distintas** (por ejemplo, una de asociación y una de comparación de grupos), con supuestos verificados e interpretación (notebooks 07-15).
5. **Modelado**: un modelo de regresión **o** una técnica multivariada (segmentación, análisis factorial, PCA), correctamente aplicada e interpretada (notebooks 16-21). Usar ambas es válido y suma en la rúbrica de criterio analítico.
6. **Comunicación**: reporte ejecutivo y presentación (ver abajo).

La *hoja de decisión* del notebook 00 es la referencia para elegir las técnicas según su pregunta y sus variables.

## Hitos y calendario

Los hitos son entregas cortas cuyo propósito es avanzar a tiempo y recibir retroalimentación; en conjunto valen el 30% de la calificación del proyecto.

| Hito | Sesión | Entregable | Peso |
|------|--------|-----------|------|
| **1. Propuesta** | Sesión 2 | Media página: integrantes, dataset y fuente, pregunta de negocio y diccionario de variables. El profesor aprueba o sugiere ajustes. | 10% |
| **2. Datos preparados + EDA** | Sesión 4 | Notebook con la limpieza documentada y justificada, medidas descriptivas y visualizaciones iniciales interpretadas. | 10% |
| **3. Análisis inferencial** | Sesión 6 | Avance del notebook con las pruebas de hipótesis: supuestos, resultados e interpretación de negocio. | 10% |
| **Entrega final** | Sesión 8 | Notebook completo + reporte ejecutivo + presentación. | 70% |

Los hitos se entregan en la plataforma del curso antes de la sesión indicada. Un hito entregado a tiempo pero con áreas de mejora recibe retroalimentación sin penalización mayor; un hito no entregado sí pierde su peso.

## Entrega final (sesión 8)

Tres componentes:

1. **Notebook completo** (ipynb exportado a html o pdf): todo el flujo, con código documentado y decisiones justificadas. Debe poder leerse de principio a fin como la historia del proyecto.
2. **Reporte ejecutivo** (máximo 2 páginas, pdf, **sin código**): dirigido a una audiencia directiva. Pregunta, hallazgos principales en lenguaje de negocio, recomendación y limitaciones. Los *ejemplos de reporte de resultados* de los notebooks del curso son la referencia de estilo.
3. **Presentación** (10 minutos + 5 de preguntas): dirigida a un "comité directivo" (el grupo y el profesor). Todos los integrantes participan. Se evalúa la claridad del mensaje, no la cantidad de láminas: qué encontraron, qué recomiendan y qué tan sólida es la evidencia.

La sesión de preguntas funciona como **defensa oral** del proyecto: cualquier integrante debe poder explicar cualquier parte del análisis, incluido el código. Un proyecto que su equipo no pueda defender se considerará evidencia de trabajo no auténtico.

## Rúbrica de la entrega final

1. **Comprensión del negocio y planteamiento (10%)**: pregunta clara, relevante y conectada con una decisión.
2. **Preparación y calidad de datos (15%)**: tratamiento técnicamente correcto y justificado; decisiones documentadas.
3. **Análisis inferencial (15%)**: pruebas apropiadas para el tipo de variables, supuestos verificados, interpretación correcta de p-valores y magnitudes.
4. **Modelado o análisis multivariado (20%)**: técnica bien seleccionada y aplicada, con evaluación de supuestos o calidad de la solución, e interpretación sólida.
5. **Comunicación: reporte ejecutivo y presentación (20%)**: hallazgos traducidos a lenguaje de negocio, narrativa clara, recomendación accionable.
6. **Criterio analítico y limitaciones (10%)**: decisiones razonadas a lo largo del flujo y reconocimiento honesto de las limitaciones del análisis.
7. **Extensión del análisis con IA Generativa (10%)**: uso de la IA para ir más allá de lo solicitado, con un anexo que incluya los *prompts* utilizados y una reflexión crítica sobre las respuestas obtenidas.

*Igual que en las prácticas: un proyecto que "corre" pero no interpreta, o que interpreta sin sustento en los datos, no alcanzará la calificación máxima. Significativo no es lo mismo que importante: la recomendación final debe considerar la relevancia práctica, no solo la estadística.*

## Errores comunes (evítenlos)

- **Empezar el proyecto en la semana 6.** Los hitos existen precisamente para evitarlo.
- **Elegir un dataset "de juguete"** ya limpio y sin contexto de negocio: hace trivial la preparación y artificial la interpretación.
- **Aplicar técnicas sin pregunta.** Primero la pregunta de negocio, luego la técnica; la hoja de decisión del notebook 00 ayuda a mapearla.
- **Reportar salidas de Python sin interpretación**, o llenar el reporte ejecutivo de tablas y código.
- **Ignorar los supuestos** de las pruebas y modelos: verificar supuestos es parte de la calificación.
- **Delegar el análisis a la IA.** Úsenla para profundizar y contrastar; las conclusiones deben reflejar comprensión propia y coherencia con sus propios resultados.

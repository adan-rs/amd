# Guía del examen final

Esta guía orienta el estudio para el examen final (sesión 8). Las preguntas buscan orientar el estudio **conceptual**: en el examen se evaluará principalmente la capacidad de **interpretar resultados, analizar escenarios y tomar decisiones** con base en estos conceptos — no la memorización de sintaxis de Python.

**Cómo estudiar**:
- Recorre los notebooks del curso con atención a las secciones *¿Para qué se utiliza?*, los supuestos y los *ejemplos de reporte de resultados*.
- Usa la **hoja de decisión** del notebook 00 para practicar la selección de técnica según el tipo de variables y el objetivo.
- Responde las preguntas de autoevaluación de los notebooks 07, 19 y 21.
- El formato y la duración del examen se anunciarán en clase.
